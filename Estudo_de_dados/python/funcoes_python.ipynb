```python
salarios = [1290, 3250, 2890, 4130, 1540]
```

map(função, iterável)

```python
def vezes_dc (x):
    y = x * 1.25
    return y
```

```python
salarios_atual = map(vezes_dc, salarios)
```

```python
print(list(salarios_atual))
```

```python
salarios_atual = map(lambda x: x * 1.25, salarios)
```

```python
print(list(salarios_atual))
```
